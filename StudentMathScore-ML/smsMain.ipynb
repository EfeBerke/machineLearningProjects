```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("student-math.csv", sep=';')

plt.xlabel("Final Grade (G3)")
plt.ylabel("Count")

df[df["G3"] > 0]["G3"].hist(bins=20)
plt.title("G3 Distribution (Zeros Removed)")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["studytime"], df["G3"])
plt.xlabel("Study Time")
plt.ylabel("Final Grade")
plt.title("Study Time vs G3")
plt.show()

# Correlation between study time and final grade is ~ 0.1 which indicates weak linear relationship
df["studytime"].corr(df["G3"]) 
```

```python
df.groupby("studytime")["G3"].mean()
# Study time alone is not a stong predictor of final grade
```

```python
df.groupby("failures")["G3"].mean()
failure_g3_corr = df["failures"].corr(df["G3"])
studytime_g3_corr = df["studytime"].corr(df["G3"])
print("Study Time - G3 Correlation: " + str(studytime_g3_corr))
print("Failure - G3 Correlation: " + str(failure_g3_corr))
```

```python
numeric_cols = df.select_dtypes(include='number')

corrs = numeric_cols.corr()["G3"].sort_values(ascending=False)

print(corrs) # Listed all correlations
```

```python
# Identifying multicollinearity
print(df["Medu"].corr(df["Fedu"]))
# 0.62 suggests that Medu and Fedu partially measuring the same thing
```

```python
# Identifying Whether Higher Education Expectation affects the G3 Score
df.groupby("higher")["G3"].mean()
# There is 3.8 difference so yes it affects!
```
